## Pregunta del proyecto

¿Cómo se relaciona el número de errores detectados en un archivo de carga con el tiempo necesario para corregirlo?

## Origen de los datos

Los datos de este notebook son sintéticos: fueron creados únicamente para esta actividad. Simulan archivos de carga con distintos números de errores y no incluyen información personal, confidencial ni datos reales de una empresa.

## Resultado esperado

El notebook creará una tabla sencilla con el número de errores y los minutos de corrección. Después calculará un promedio y mostrará una gráfica para observar si, al aumentar los errores, también aumenta el tiempo de corrección.

```python
import sys
import pandas as pd
import matplotlib.pyplot as plt

print("Versión de Python:", sys.version)
print("Paquetes preparados correctamente.")
```

```python
datos = {
    "archivo": ["Archivo A", "Archivo B", "Archivo C", "Archivo D", "Archivo E", "Archivo F"],
    "errores_detectados": [1, 2, 3, 4, 5, 6],
    "minutos_correccion": [5, 9, 14, 20, 26, 32]
}

tabla = pd.DataFrame(datos)

tabla
```

```python
promedio_tiempo = tabla["minutos_correccion"].mean()

print(f"El tiempo promedio de corrección es: {promedio_tiempo:.1f} minutos")

plt.figure(figsize=(8, 5))
plt.plot(
    tabla["errores_detectados"],
    tabla["minutos_correccion"],
    marker="o",
    color="#2F6B8A"
)

plt.title("Errores detectados vs. tiempo de corrección")
plt.xlabel("Número de errores detectados")
plt.ylabel("Minutos de corrección")
plt.grid(True, alpha=0.3)
plt.show()
```

## Conclusión

En este ejemplo sintético se observa que, conforme aumenta el número de errores detectados en un archivo, también aumenta el tiempo necesario para corregirlo.

El tiempo promedio de corrección fue de 17.7 minutos. Esto sugiere que validar los archivos antes de revisarlos manualmente puede reducir el retrabajo y hacer más eficiente el proceso.
